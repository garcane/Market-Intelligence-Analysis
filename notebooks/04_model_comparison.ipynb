Stage 9 baseline model results, loaded from the trained artifacts (`outputs/model_results/models/`) rather than retrained here — training happens once via `python -m src.models.train_baselines`; see `MODELS.md` for the full writeup including the overfitting-then-regularization story.

```python
import sys, os, json
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from src.config import PROCESSED_DIR

with open(PROCESSED_DIR / 'model_report_h5d.json') as f:
    report = json.load(f)
print('Leaderboard (by', report['leaderboard_criteria'], '):', report['leaderboard'])
print('Any suspiciously high AUC (>=0.90)?', report['any_suspiciously_high_auc'])
```

## Validation metrics, all models

```python
rows = []
for name, m in report['models'].items():
    vm = m.get('validation_metrics', {})
    rows.append({'model': name, **{k: v for k, v in vm.items() if k != 'confusion_matrix'}})
pd.DataFrame(rows).set_index('model')
```
